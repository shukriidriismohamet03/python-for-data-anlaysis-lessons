```python
import pandas as pd
```

```python
# loading data
data=pd.read_excel("/content/drive/MyDrive/Colab Notebooks/CRIME-DATA-ANALYSIS-COMPLA.xlsx")
```

```python
data.head()
```


##Renaming

```python
data = data.rename(columns={"Crime_ID": "Crimid"})
```

```python
data.head()
```

```python
data["District"].value_counts()
```

```python
data["District"] = data["District"].replace("hodan", "Hodan")
```

```python
data["District"].value_counts()
```

```python
data["Crime_Type"].value_counts()
```

```python
data["Victim_Gender"] = data["Victim_Gender"].replace({
    "male": "Male",
    "female": "Female"
})
```

```python
data["Victim_Gender"].value_counts()
```

```python
data["Case_Status"] = data["Case_Status"].replace({
    "open": "Open",
    "closed": "Closed",
    "under investigation": "Under Investigation"
})
```

```python
data["Case_Status"].value_counts()
```

##numerical replacment

```python
data["Severity_Level"] = data["Severity_Level"].replace({
    "Low": 1,
    "Medium": 2,
    "High": 3
})
```

```python
data["Severity_Level"].value_counts()
```
